-----------------   Section A – Conceptual Understanding  -----------------


A1. Why are NumPy arrays preferred over Python lists?
NumPy arrays are generally faster than normal Python lists when we work with numerical data. They also use memory more efficiently because the elements in an array usually have the same data type. NumPy provides useful mathematical operations such as mean, standard deviation, matrix multiplication, and aggregation directly on arrays. 


A2. What is vectorization?

Vectorization means performing an operation on an entire NumPy array at once instead of processing each element using a Python for loop. For example, we can multiply an entire array by 2 using arr * 2. This is important for large datasets because NumPy's optimized operations are usually much faster than running Python loops element by element.



A3. Difference between A * B and A @ B

A * B performs element-wise multiplication, meaning corresponding elements of the two arrays are multiplied Where, A @ B performs matrix multiplication, where rows of one matrix are multiplied with columns of another matrix. 

A4. What is broadcasting?

Broadcasting is a NumPy feature that allows arrays with different but compatible shapes to be used together in element-wise operations. NumPy compares their dimensions from right to left. Two dimensions are compatible when they are equal, or when one of them is 1; missing dimensions are also treated as 1. This allows operations such as adding a single value or a row/column vector to a larger array.

A5. flatten() vs ravel()

Both flatten() and ravel() convert an array into a one-dimensional array. The main difference is that flatten() always creates a copy, while ravel() tries to return a view of the original array whenever possible. In a memory-constrained application processing large images, I would prefer ravel() because it can avoid creating an unnecessary copy and therefore save memory.

A6. Explain axis=0 and axis=1
For a 2D array, axis=0 means the operation is performed down the rows, so we get a result for each column. axis=1 means the operation is performed across the columns, so we get a result for each row.

```python
import numpy as np
a = np.array([[1, 2, 3],[4, 5, 6]])
print(np.sum(a, axis=0))
print(np.sum(a, axis=1))
```

A7. Normalization vs Standardization

Normalization and standardization are used to bring numerical features into a suitable range or scale before training an ML model. Normalization usually changes values to a fixed range, commonly [0, 1], using the minimum and maximum values. Standardization changes values so that the mean becomes approximately 0 and the standard deviation becomes 1. Scaling is especially useful for models where features with larger numerical values could otherwise dominate smaller-valued features.

A8. Why does np.nanmean() exist?

np.nanmean() is useful when our data contains NaN values and we want to calculate the mean while ignoring those missing values. If we use np.mean() directly on an array containing NaN, the result will normally become NaN. Therefore, np.nanmean() is useful when missing values are present and we want the mean of the available values.

A10. Why is reproducibility important?

Reproducibility means getting the same results when an experiment is repeated under the same conditions. Random operations can produce different results each time, which makes it difficult to compare experiments.

C1. Given a 1D NumPy array of exam scores, use np.where() to produce labels 'P' (score >= 40) or 'F' (score < 40).

C2.Write a function standardize(x) that takes a 1D NumPy array and returns its z-score standardized version using only NumPy.

```python
import numpy as np
def standardize(x):
    mean = np.mean(x)
    std = np.std(x)
    return (x - mean) / std
x = np.array([10, 20, 30, 40, 50])
result = standardize(x)
print(result)

# The z-score is calculated using (x - mean) / standard deviation, which centers the data around 0.
```

C5. Implement y_pred = X @ w + b for a feature matrix X, weight vector w and bias b. Then compute MSE using y_true. 

```python
import numpy as np
X = np.array([[1, 2],[2, 3],[3, 4]])
w = np.array([0.5, 1.0])
b = 2.0
y_true = np.array([4, 6, 8])
y_pred = X @ w + b
mse = np.mean((y_true - y_pred) ** 2)
print("Predicted:", y_pred)
print("MSE:", mse)

#  X @ w + b calculates the predictions, and MSE calculates the average squared difference between actual and predicted values.
```

C7. Given a grayscale image as a 2D NumPy array with pixel values in [0, 255], normalize the values to [0, 1] using min-max normalization.

```python
import numpy as np
image = np.array([[0, 50, 100],[150, 200, 255]])
normalized = (image - image.min()) / (image.max() - image.min())
print(normalized)

# Min-max normalization converts the minimum pixel value to 0 and the maximum pixel value to 1, with the other values scaled between them.
```

D1.Data Preprocessing Pipeline 
You are given a raw 1D or 2D NumPy array containing NaN values.
Describe, step by step, the NumPy operations you would apply to:
(a) detect missing values,
(b) handle them appropriately and justify your strategy, and
(c) normalize or standardize the clean data before it is fed to a model.


```python
import numpy as np
data = np.array([[10.0, 100.0],[20.0, np.nan],[30.0, 300.0],[np.nan, 400.0]])

missing = np.isnan(data)
print("Missing values:")
print(missing)

column_mean = np.nanmean(data, axis=0)

# Replaced NaN values with column mean
clean_data = np.where(missing, column_mean, data)
print("\nClean data:")
print(clean_data)

mean = clean_data.mean(axis=0)
std = clean_data.std(axis=0)
standardized = (clean_data - mean) / std
print("\nStandardized data:")
print(standardized)

# np.isnan() detects missing values, np.nanmean() helps calculate replacement values while ignoring NaN, and the cleaned data is standardized using mean and standard deviation.
```

E1. Per-Column Min–Max Normalization [5 marks]
Implement min-max normalization for a 2D feature matrix on a per-column (per-feature) basis, rather than using a single global min/max for the whole array. Your solution must be fully vectorized, with no explicit loops over rows or columns.

```python
import numpy as np
X = np.array([[10, 100],[20, 200],[30, 300],[40, 400]])
column_min = X.min(axis=0)
column_max = X.max(axis=0)
normalized = (X - column_min) / (column_max - column_min)
print("Normalized data:")
print(normalized)

# axis=0 calculates the minimum and maximum separately for each feature column, and broadcasting applies the normalization without using loops.
```

E2. Three-Level Score Classification [5 marks]
Given a 1D array of exam scores, use a single vectorized expression (no if/else statements and no explicit loops) to classify each score as:
Fail (< 40), Average (40–70 inclusive), or Excellent (> 70).

Hint: You may need to nest or combine np.where() calls.

```python
import numpy as np
scores = np.array([25, 40, 55, 70, 71, 90])
result = np.where(
    scores < 40,
    "Fail",
    np.where(scores <= 70, "Average", "Excellent"))
print(result)

# The first np.where() identifies scores below 40 as Fail, and the nested np.where() classifies the remaining scores as Average or Excellent.
```
